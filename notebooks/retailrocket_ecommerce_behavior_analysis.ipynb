```python
# difference between consecutive events for the same visitor 
df["time_diff"] = (df.groupby("visitorid")["timestamp"].diff() / 1000 / 60)
```

```python
print("Time gap statistics (minutes):")
display(df["time_diff"].describe())
```

```python
#how many gaps exceed 30 minutes
print("Gaps above 30 minutes:",(df["time_diff"] > 30).sum())
```

```python
#percentage
print("Percentage:",round((df["time_diff"] > 30).mean() * 100, 2),"%")
```

```python
df["new_session"] = (df["time_diff"].isna() |(df["time_diff"] > 30))
```

```python
df["session_number"] = (df.groupby("visitorid")["new_session"].cumsum())
```

```python
# Create unique session ID
df["session_id"] = (df["visitorid"].astype(str) + "_" + df["session_number"].astype(str))
```

```python
#event types available in each session
session_events = (df.groupby("session_id")["event"].agg(list))
```

```python
session_events
```

```python
# Count sessions containing each event
print("Sessions containing each event:")
for event in ["view", "addtocart", "transaction"]:
    count = session_events.apply(lambda events: event in events).sum()
    print(event, ":", count)
```

```python
is_view = df["event"].eq("view")
```

```python
is_cart = df["event"].eq("addtocart")
```

```python
is_transaction = df["event"].eq("transaction")
```

```python
df["view_seen"] = (is_view.groupby(df["session_id"]).cummax())
```

```python
df["cart_after_view"] = ((is_cart & df["view_seen"]).groupby(df["session_id"]).cummax())
```

```python
df["transaction_after_cart"] = (is_transaction & df["cart_after_view"])
```

```python
funnel_view = df.loc[is_view, "session_id"].nunique()
```

```python
funnel_view
```

```python
funnel_cart = df.loc[ is_cart & df["cart_after_view"], "session_id"].nunique()
```

```python
funnel_cart
```

```python
funnel_transaction = df.groupby("session_id")["transaction_after_cart"].any().sum()
```

```python
funnel_transaction
```

```python
print("View:", funnel_view)
print("Add to Cart after View:", funnel_cart)
print("Transaction after Cart:", funnel_transaction)
```

```python
view_to_cart = funnel_cart / funnel_view * 100
```

```python
cart_to_transaction = funnel_transaction / funnel_cart * 100
```

```python
overall_conversion = funnel_transaction / funnel_view * 100
```

```python
view_dropoff = 100 - view_to_cart
```

```python
cart_dropoff = 100 - cart_to_transaction
```

```python
print("View to Cart:", view_to_cart)
print("Cart to Transaction:", cart_to_transaction)
print("Overall Conversion:", overall_conversion)
```

```python
print("View Drop-off:", view_dropoff)
print("Cart Drop-off:", cart_dropoff)
```

# Retention Analysis

```python
first_activity = (df.groupby("visitorid")["datetime"].min())

first_activity.head()
```

```python
active_days = (df.groupby("visitorid")["date"].nunique())

active_days.head()
```

```python
active_days.value_counts().sort_index()
```

```python
total_users = active_days.shape[0]
```

```python
total_users
```

```python
returning_users
```

```python
returning_rate = returning_users / total_users * 100
```

# Key Business Insights

1. Funnel Conversion: Only 2.09% of sessions with product views proceeded to add-to-cart events.
2. Checkout Progression: 29.23% of sessions with an add to cart after a view proceeded to a transaction event.
3. User Retention: The observed multi day activity rate was 10.21%, indicating that most visitors were active on only one observed calendar day.
4. Cohort Retention: May 2015 users had a 4.76% monthly retention rate in the following month.
5. Session Engagement: The median number of events per session was one, suggesting limited event activity in most sessions
